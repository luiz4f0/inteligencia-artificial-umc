```python
# 21- so funciona em lista JA ordenada
lista = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
alvo = int(input("Qual numero procurar?"))
inicio = 0
fim = len(lista) - 1
posicao = -1   # -1 significa "nao achei ainda"
while inicio <= fim:
    meio = (inicio + fim) // 2   # // e divisao inteira
    if lista[meio] == alvo:
        posicao = meio
        break
    elif lista[meio] < alvo:
        inicio = meio + 1   # procura na metade da direita
    else:
        fim = meio - 1      # procura na metade da esquerda
print("Posicao:", posicao)
```

```python
# 22- ""letra.isalpha" so mexe se for letra,  65 e o codigo do A maiusculo, 97 do a minusculo
# "resultado = resultado + letra" espaco e pontuacao passam direto
# "ord() vira numero, chr() volta pra letra, % 26 da a volta no alfabeto
mensagem = input("Mensagem: ")
chave = int(input("Deslocamento"))
resultado = ""
for letra in mensagem:
    if letra.isalpha():
        base = 65 if letra.isupper() else 97
        resultado = resultado + chr((ord(letra) - base + chave) % 26 + base)
    else:
        resultado = resultado + letra
print("Cifrada:", resultado)
```

```python
# 23- mesma coisa da 22, só muda o menos na frente da chave (volta o alfabeto)
mensagem = input("Mensagem cifrada:")
chave = int(input("Deslocamento usado:"))
resultado = ""
for letra in mensagem:
    if letra.isalpha():
        base = 65 if letra.isupper() else 97
        resultado = resultado + chr((ord(letra) - base - chave) % 26 + base)
    else:
        resultado = resultado + letra
print("Original:", resultado)
```

```python
# 24- O "cpf.isdigit" pergunta: essa string tem só números
cpf = input("Digite o CPF (so numeros):")
valido = True
if len(cpf) != 11 or not cpf.isdigit():
    valido = False
elif cpf == cpf[0] * 11:
    valido = False
else:
    for i in range(9, 11):
        soma = 0
        for j in range(i):
            soma = soma + int(cpf[j]) * (i + 1 - j)
        digito = (soma * 10 % 11) % 11
        if digito == 10:
            digito = 0
        if digito != int(cpf[i]):
            valido = False
if valido:
    print("CPF valido!")
else:
    print("CPF invalido.")
```

```python
# 30 - Jogo da forca: o usuario chuta letras ate acertar a palavra
# random.choice = sorteia um item de dentro da lista
# acertos = lista com as letras que o jogador ja acertou
# exibicao = a palavra montada com _ no lugar do que ainda falta
# in = pergunta se a letra esta dentro da palavra
# o laco roda enquanto os erros forem menos de 6
import random
palavra = random.choice(["python", "teclado", "computador"])
acertos = []
erros = 0
while erros < 6:
    exibicao = ""
    for letra in palavra:
        if letra in acertos:
            exibicao = exibicao + letra
        else:
            exibicao = exibicao + "_"
    print(exibicao, "| erros:", erros)
    if "_" not in exibicao:
        print("Voce venceu!")
        break
    chute = input("Chute uma letra: ").lower()
    if chute in palavra:
        acertos.append(chute)
    else:
        erros = erros + 1
if erros == 6:
    print("Perdeu! A palavra era:", palavra)
```

```python
# 31 - Ordena a lista comparando vizinhos e trocando de lugar
# n = len(lista) guarda o tamanho da lista
# o for de fora conta quantas passadas faremos pela lista
# o for de dentro compara cada item com o vizinho da direita
# lista[j], lista[j+1] = lista[j+1], lista[j] troca os dois de lugar
lista = [64, 34, 25, 12, 22, 11, 90]
n = len(lista)
for i in range(n):
    for j in range(n - i - 1):
        if lista[j] > lista[j + 1]:
            lista[j], lista[j + 1] = lista[j + 1], lista[j]
print(lista)
```

```python
5# 32 - Jogo da velha para dois jogadores no mesmo teclado
# [" "] * 9 = cria 9 casas vazias (posicoes 0 a 8)
# vitorias = todas as combinacoes de casas que fecham o jogo
# continue = pula pra proxima volta do laco sem executar o resto
# != " " garante que a combinacao vencedora nao seja de casas vazias
# a ultima linha troca a vez do jogador
tabuleiro = [" "] * 9
jogador = "X"
vitorias = [[0,1,2], [3,4,5], [6,7,8],
            [0,3,6], [1,4,7], [2,5,8],
            [0,4,8], [2,4,6]]
for rodada in range(9):
    for i in range(0, 9, 3):
        print(tabuleiro[i], "|", tabuleiro[i+1], "|", tabuleiro[i+2])
    pos = int(input("Jogador " + jogador + ", escolha (0-8): "))
    if tabuleiro[pos] != " ":
        print("Casa ocupada!")
        continue
    tabuleiro[pos] = jogador
    venceu = False
    for c in vitorias:
        if tabuleiro[c[0]] == tabuleiro[c[1]] == tabuleiro[c[2]] != " ":
            venceu = True
    if venceu:
        print("Jogador", jogador, "venceu!")
        break
    jogador = "O" if jogador == "X" else "X"
```

```python
# 37 - Calcula o fatorial usando recursao (funcao que chama ela mesma)
# def = cria uma funcao / o nome dela aqui e fatorial
# n = o valor que a funcao recebe pra trabalhar
# return = devolve o resultado pra quem chamou a funcao
# o if n <= 1 e o caso base: a porta de saida, senao rodaria pra sempre
# este exercicio pede funcao, por isso o def e obrigatorio aqui
def fatorial(n):
    if n <= 1:
        return 1
    return n * fatorial(n - 1)
numero = int(input("Digite um numero: "))
print("Fatorial:", fatorial(numero))
```

```python
# 40 - Menu para adicionar, remover e listar itens de compras
# lista = [] comeca vazia
# while True = fica repetindo o menu ate o usuario escolher sair
# .append = adiciona um item / .remove = tira um item
# o if item in lista evita erro ao remover algo que nao existe
# break = encerra o laco e termina o programa
lista = []
while True:
    print("1-Adicionar  2-Remover  3-Listar  4-Sair")
    opcao = input("Escolha:")
    if opcao == "1":
        lista.append(input("Item:"))
    elif opcao == "2":
        item = input("Item a remover:")
        if item in lista:
            lista.remove(item)
        else:
            print("Nao encontrado.")
    elif opcao == "3":
        for item in lista:
            print("-", item)
    elif opcao == "4":
        break
```

```python
# 46 - Calcula a media e o desvio padrao de uma lista
# media = soma dividida pela quantidade de itens
# o for soma o quanto cada numero se afasta da media, elevado ao quadrado
# variancia = essa soma dividida pela quantidade de itens
# math.sqrt = raiz quadrada, que transforma a variancia em desvio padrao
import math
numeros = [10, 12, 23, 23, 16, 23, 21, 16]
media = sum(numeros) / len(numeros)
soma = 0
for n in numeros:
    soma = soma + (n - media) ** 2
variancia = soma / len(numeros)
desvio = math.sqrt(variancia)
print("Media:", media)
print("Desvio padrao:", desvio)
```

```python
# 48 - Ordena a lista encaixando cada item na posicao certa
# funciona como organizar cartas na mao, uma por vez
# comeca do indice 1 porque o primeiro item sozinho ja esta ordenado
# atual = o item que estamos segurando pra encaixar
# o while empurra pra direita todo mundo maior que o item atual
# no fim, o item atual entra no buraco que sobrou
lista = [12, 11, 13, 5, 6]
for i in range(1, len(lista)):
    atual = lista[i]
    j = i - 1
    while j >= 0 and lista[j] > atual:
        lista[j + 1] = lista[j]
        j = j - 1
    lista[j + 1] = atual
print(lista)
```
