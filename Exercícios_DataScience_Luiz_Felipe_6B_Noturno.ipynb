```python
# SciPy
import numpy as np
from scipy import integrate, linalg, optimize, interpolate
from scipy.fft import fft, ifft
np.random.seed(123)
```

```python
# Exercício 1
f = lambda x: x**2 * np.sin(x)
r, e = integrate.quad(f, 0, np.pi)
print(r)
```

```python
# Exercício 2
edo = lambda x, y: x - y
sol = integrate.solve_ivp(edo, [0, 10], [1], t_eval=np.linspace(0, 10, 6))
print(sol.t)
print(sol.y[0])
```

```python
# Exercício 3
raizes = np.roots([1, -3, 0, 0, 2])
print(raizes)
```

```python
# Exercício 4
res = optimize.minimize_scalar(lambda x: -(-x**2 + 4*x + 1))
print(res.x, -res.fun)
```

```python
# Exercício 5
t = np.linspace(0, 1, 400, endpoint=False)
sinal = np.sin(2*np.pi*t) + np.cos(4*np.pi*t)
print(np.abs(fft(sinal))[:8])
```

```python
# Exercício 6
A = np.random.rand(5, 5)
inv = linalg.inv(A)
print(np.round(A @ inv, 6))
```

```python
# Exercício 7
x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 3, 5, 4])
cs = interpolate.CubicSpline(x, y)
print(cs([0.5, 1.5, 2.5, 3.5]))
```

```python
# Exercício 8
A = np.array([[3, 2], [1, 4]])
b = np.array([5, 6])
print(linalg.solve(A, b))
```

```python
# Exercício 9
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])
print(np.polyfit(x, y, 2))
```

```python
# Exercício 10
r, e = integrate.dblquad(lambda y, x: x*y + x**2, 0, 2, 0, 1)
print(r)
```

```python
# Exercício 11
M = np.random.randint(1, 10, (4, 4)).astype(float)
w, v = linalg.eig(M)
print(w)
```

```python
# Exercício 12
g = lambda x: np.exp(x) - x**2
print(optimize.newton(g, 1))
```

```python
# Exercício 13
M = np.random.uniform(-10, 10, (6, 6))
print(linalg.det(M))
```

```python
# Exercício 14
f = lambda p: (p[0]-2)**2 + (p[1]-3)**2
res = optimize.minimize(f, [5, 5])
print(res.x)
```

```python
# Exercício 15
s = np.random.rand(8) + 1j*np.random.rand(8)
print(ifft(s))
```

```python
# Exercício 16
def sistema(x, estado):
    u, v = estado
    return [v, -u]
sol = integrate.solve_ivp(sistema, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 6))
print(sol.y[0])
```

```python
# Exercício 17
print(optimize.brentq(lambda x: x**3 - 27, 0, 10))
```

```python
# Exercício 18
x = np.array([1, 2, 3, 4, 5])
y = x**2
f = interpolate.interp1d(x, y, kind='cubic')
print(f([1.5, 2.5, 3.5, 4.5]))
```

```python
# Exercício 19
n = 64
t = np.linspace(0, 1, n, endpoint=False)
s = np.sin(2*np.pi*5*t) + np.cos(2*np.pi*10*t)
print(np.abs(fft(s))[:12])
```

```python
# Exercício 20
A = np.array([[1, 1, 1], [0, 2, 5], [2, 5, -1]])
b = np.array([6, -4, 27])
print(linalg.solve(A, b))
```

```python
# Pandas
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def base():
    return pd.DataFrame({
        "Produto":    ["Monitor", "Bateria", "Teclado", "Notebook", "Cabo"],
        "Preço":      [899.90, 45.00, 120.00, 3200.00, 19.90],
        "Quantidade": [15, 80, 40, 8, 200],
        "Categoria":  ["Periférico", "Energia", "Periférico", "Computador", "Periférico"],
    })
```

```python
# Exercício 1
df = base()[["Produto", "Preço", "Quantidade"]]
print(df)
```

```python
# Exercício 2
base().to_excel("vendas.xlsx", index=False)
print(pd.read_excel("vendas.xlsx").head(10))
```

```python
# Exercício 3
df = base()
print(df[df["Preço"] > 100])
```

```python
# Matplotlib
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator
from mpl_toolkits.mplot3d import Axes3D
np.random.seed(7)
```
